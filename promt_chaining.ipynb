```python
from langchain_ollama  import ChatOllama
from langgraph.graph import StateGraph,START,END 
from typing import TypedDict
```

```python
class state(TypedDict):
    topic:str
    blog:str
    outline:str 
```

```python
model=ChatOllama(
    model='llama3.2:3b',
    temperature=0.7
)
def outline(state:state):
    topic=state['topic']
    prompt=f"generate the detailed outline for creating blog on following topic \n ->{topic}"
    result=model.invoke(prompt)
    return {'outline':result.content}

def blog(state:state):
    topic=state['topic']
    outline=state['outline']
    prompt=f"genrate the detailed blog for follwing topic ->{topic}and the blog should be in given outline{outline}"
    result=model.invoke(prompt)
    return {'blog':result.content}

```

```python
graph=StateGraph(state)

#defining nodes
graph.add_node('outline_genrator',outline)
graph.add_node('blog_genrator',blog)

#definig_edge
graph.add_edge(START,'outline_genrator')
graph.add_edge('outline_genrator','blog_genrator')
graph.add_edge('blog_genrator',END)

workflow=graph.compile()

intial_state={'topic':"indian youth"}
result=workflow.invoke(intial_state)
print(result['blog'])
```
